# Rotation Modelling

```python
import numpy as np
import cmath as cm
```

```python
pi = np.pi

def cart2pol(x, y, z):
    rho = np.sqrt(x**2 + y**2 + z**2)
    theta = np.arccos(z/rho)
    
    if x != 0:
        phi = np.arctan(y/x)
    else:
        if y >= 0:
            phi = 0
        elif y < 0:
            phi = np.pi
    
    return(rho, theta, phi)

def pol2cart(rho, theta, phi):
    x = rho * np.sin(theta) * np.cos(phi)
    y = rho * np.sin(theta) * np.sin(phi)
    z = rho * np.cos(theta)
    
    x = np.round(x, 3)
    y = np.round(y, 3)
    z = np.round(z, 3)
    
    return(x, y, z)

def rad2deg(rad):
    deg = rad * np.pi/180
    return deg
```
